## Tabela de frequência e gráfico da variável 'Survived'

```python

# Tabela de frequência da variável Survived
freq_survived = df['Survived'].value_counts()

print(freq_survived)

# Gráfico
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', data=df)
plt.title('Frequência de Sobreviventes')
plt.xlabel('Sobreviveu')
plt.ylabel('Quantidade')
plt.show()
```


### Explicação do gráfico
Foi utilizado um gráfico de barras porque a variável **Survived** é categórica (0 = não sobreviveu, 1 = sobreviveu). 
Esse tipo de gráfico facilita a comparação entre as quantidades de cada categoria.


## Quantas pessoas sobreviveram? Qual a classe mais frequente?

```python

# Quantidade de sobreviventes
sobreviventes = df['Survived'].sum()

print(f'Quantidade de sobreviventes: {sobreviventes}')

# Classe mais frequente
classe_mais_frequente = df['Pclass'].mode()[0]

print(f'Classe mais frequente: {classe_mais_frequente}')
```

## Relação entre sexo e sobrevivência

```python

# Tabela de frequência entre sexo e sobrevivência
tabela_sexo = pd.crosstab(df['Sex'], df['Survived'])

print(tabela_sexo)

# Gráfico
plt.figure(figsize=(6,4))
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Relação entre Sexo e Sobrevivência')
plt.xlabel('Sexo')
plt.ylabel('Quantidade')
plt.show()
```


### Análise
Observando a tabela e o gráfico, percebe-se que as mulheres sobreviveram proporcionalmente mais do que os homens.


## Distribuição das idades dos passageiros

```python

# Gráfico da distribuição das idades
plt.figure(figsize=(8,5))
sns.histplot(df['Age'].dropna(), bins=30, kde=True)

plt.title('Distribuição das Idades dos Passageiros')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.show()
```


### Análise da distribuição
A distribuição das idades não parece perfeitamente simétrica, apresentando maior concentração de passageiros jovens e adultos.
A maioria dos passageiros era jovem, com menos pessoas idosas no conjunto de dados.
